```python
# Load env variables
from dotenv import load_dotenv

import os

# Option 2: safer — won't error if it's already unset
os.environ.pop("ANTHROPIC_API_KEY", None)

# Load API key from .env file (expects ANTHROPIC_API_KEY)
load_dotenv()
```

```python
# Create an API client
import anthropic

# Initialize the client (automatically uses ANTHROPIC_API_KEY from environment)
client = anthropic.Anthropic()

#model = "claude-sonnet-4-0"
model = "claude-sonnet-4-5-20250929" # Not sooner than September 29, 2026

# print the api key (for debugging)
#import os
#print(os.environ.get("ANTHROPIC_API_KEY"))
#print(len(os.environ.get("ANTHROPIC_API_KEY")))
#print(repr(os.environ.get("ANTHROPIC_API_KEY")))
```

```python
# Make a request
message = client.messages.create(
    model=model,
    max_tokens=10,
    messages=[
        {
            "role" : "user",
            "content" : "What is Quantum Computing? Answer in one sentence."
        }
    ]
)
```
